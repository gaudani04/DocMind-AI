## Embeddings

```python
import numpy as np
from sentence_transformers import SentenceTransformer
import chromadb
from chromadb.config import Settings
import uuid
from typing import List,Dict,Any,Tuple
from sklearn.metrics.pairwise import cosine_similarity
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
## chunks are document so we need to extract text first
texts = [i.page_content for i in chunks]
embeddings = model.encode(texts)
```

```python
len(embeddings)
```

```python
import faiss
import numpy as np

dimension = embeddings.shape[1]
index = faiss.IndexFlatL2(dimension)
index.add(np.array(embeddings))
```

```python
query = "What is Machine Learning?"

query_embedding = model.encode([query])

D, I = index.search(query_embedding, k=3)
```

```python
for i in I[0]:
    print(chunks[i].page_content)
```

```python
from chromadb.utils import embedding_functions

client = chromadb.PersistentClient(path="./EmbeddingData/vectore_store")
collection = client.get_or_create_collection(name="text_collection")
```

```python
batch_size = 16

for i in range(0, len(chunks), batch_size):
    collection.add(
        documents=[chunk.page_content for chunk in chunks[i:i+batch_size]],
        embeddings=embeddings[i:i+batch_size].tolist(),
        ids=[str(j) for j in range(i, i+len(chunks[i:i+batch_size]))]
    )
```

```python
import chromadb
from sentence_transformers import SentenceTransformer
from langchain_community.llms import Ollama

model = SentenceTransformer("all-MiniLM-L6-v2")
llm = Ollama(model="gemma3:1b")

client = chromadb.PersistentClient(path="./EmbeddingData/vectore_store")
collection = client.get_collection(name="text_collextion")

def ask_question(query):

    query_embedding = model.encode([query]).tolist()

    results = collection.query(
        query_embeddings=query_embedding,
        n_results=5
    )

    context = " ".join(results["documents"][0])

    prompt = f"""
    Use the context to answer the question.

    Write a clear paragraph summary.
    Do NOT copy sentences or references.

    Context:
    {context}

    Question:
    {query}
    """

    response = llm.invoke(prompt)

    return response
```

```python
ask_question(query="What is figuring out?")
```
